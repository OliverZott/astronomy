# Units and Constants

```python
import astropy.constants as const
import astropy.units as u
import numpy as np
```

```python
# 1. Quantities = value * unit
d = 1 * u.pc
print(d.unit)
print(d)
print(d.to(u.lyr))
print(f"{d.to(u.km).value:.3e}")
```

```python
# 2. Arithmetics combine units automatically
v = 3 * 10**5 * u.km / u.s
t = 8 * u.min
d = v * t
print(d.to(u.AU))
print(d)
print(d.decompose())
```

```python
# 3. Constants are Quantities too
# const.M_sun, const.R_earth, const.c
print(const.G)  # value, unit, uncertainty, reference
```

```python
# 4 Example: earth's escape velocity v = sqrt(2GM/R)
v_esc = np.sqrt(2 * const.G * const.M_earth / const.R_earth)
print(v_esc)
print(v_esc.to(u.km / u.s))

# 5. Example: Schwarzschild radius of the Sun, r_s = 2GM/c^2
r_s = 2 * const.G * const.M_sun / const.c**2
print(r_s.to(u.km))
```

```python
# 6. Dimension errors are caught
try:
    (1 * u.m).to(u.s)
except u.UnitConversionError as e:
    print(e)
```

```python
# 7. Equivalencies: conversions that need physics, not just algebra
print((500 * u.nm).to(u.THz, equivalencies=u.spectral()))  # wavelength to frequency
print((1 * u.arcsec).to(u.pc, equivalencies=u.parallax()))  # parallax to distance = 1 pc
```

```python
# 8. Example: Compute the orbital period of Earth with Kepler's third law, T = 2π·sqrt(a³ / (G·M_sun)),
# using a = 1 AU, and convert it to days. You should get about 365.25 d.

a = 1 * u.AU
T = 2 * np.pi * np.sqrt(a**3 / (const.G * (const.M_sun + const.M_earth)))
print(T.to(u.d))
```
