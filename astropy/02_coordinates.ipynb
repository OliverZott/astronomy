# Sky Coordinates



```python
import astropy.units as u
from astropy.coordinates import SkyCoord
```

```python
vega = SkyCoord(ra=279.23473479 * u.deg, dec=-16.71611586 * u.deg, frame="icrs")

sirius = SkyCoord(
    ra=101.28715533 * u.deg,
    dec=-16.71611586 * u.deg,
    frame="icrs",
)

print("Vega RA:", vega.ra.to_string(unit=u.hourangle, sep=":", precision=2))

separation = vega.separation(sirius)
print(f"Angular separation: {separation.to_value(u.deg):.2f} deg")
```

```python
# 18 36 56.33635, +38 47 01.2802: hours/minutes/seconds for RA and degrees/arcminutes/arcseconds for Dec.
# 279.23473479°, +38.78368894°: decimal degrees for both.

# 1 hour of RA equals 15°

ra = "18 36 56.33635"  # replace with the RA you copied
dec = "+38 47 01.2802"  # replace with the signed Dec you copied

vega = SkyCoord(
    ra=ra,
    dec=dec,
    unit=(u.hourangle, u.deg),
    frame="icrs",
)

print(vega)
print(vega.to_string("hmsdms", precision=3))

# HMS to decimal degrees
# convert to decimal hours first then times 15 to convert deciaml hour to degree
ra_decimal_hours = 18 + 36 / 60 + 56.33635 / 3600
ra_deg = ra_decimal_hours * 15

# DMS to decimal degrees
dec_deg = 38 + 47 / 60 + 01.2802 / 3600
print(f"ra in deg = {ra_deg}, dec in deg = {dec_deg}")
```
