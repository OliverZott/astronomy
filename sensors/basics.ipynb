```python
# Examples

# speed of light in m/s
c = 3.0e8  # 3.0 × 10^8

# wavelength in meters (780 nm)
wavelength = 780e-9  # 780 × 10^-9

# frequency calculation (m/s)
frequency = c / wavelength

frequency2 = 1e9

print(f"Frequency: {frequency:.3e} Hz")


def frequency_unit_converter(value: float):
    units = ["KHz", "MHz", "GHz", "THz", "PHz", ]
    index = 0

    while value >= 1000:
        value /= 1000
        index += 1

    unit = units[index]

    return f"Frequency: {value:.3e} {unit}"


print(frequency_unit_converter(frequency))
print(frequency_unit_converter(frequency2))
```
